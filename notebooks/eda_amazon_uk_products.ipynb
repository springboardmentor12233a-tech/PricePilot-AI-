```python
display(Markdown("### 📥 Loading Full Dataset"))

df6 = pd.read_csv(
    r"C:\Users\sobhi\Downloads\PRICEPIOLET AI\data\raw\amazon_uk_products_2023\amz_uk_processed_data.csv",
    low_memory=False
)

print("Shape (full dataset):", df6.shape)
df6.head()
```

```python
display(Markdown("### 🔍 Step 1: Data Types & Structure"))
df6.info()
```

```python
display(Markdown("### ⚠️ Step 2: Missing Values Check"))
print(df6.isnull().sum())
```

```python
display(Markdown("### 🔁 Step 3: Duplicate Check"))
print("Duplicate rows:", df6.duplicated().sum())
```

```python
display(Markdown("### 📈 Step 4: Summary Statistics"))
df6.describe()
```

```python
display(Markdown("### ✅ Step 5: Validity Checks"))
print("Negative price:", (df6['price'] < 0).sum())
print("Zero price:", (df6['price'] == 0).sum())
print("Negative reviews:", (df6['reviews'] < 0).sum())
print("Negative stars:", (df6['stars'] < 0).sum())
print("Stars > 5:", (df6['stars'] > 5).sum())
print("Unique categoryName:", df6['categoryName'].nunique())
print("isBestSeller values:", df6['isBestSeller'].unique())
print("Products with price > 1000:", (df6['price'] > 1000).sum())
```

```python
plt.figure(figsize=(8,5))
df6['categoryName'].value_counts().head(10).plot(kind='barh', color='steelblue')
plt.title('Top 10 Categories by Product Count')
plt.xlabel('Number of Products')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df6['price'], bins=50, range=(0,200), color='coral')
plt.title('Price Distribution (under £200)')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df6.sample(3000), x='price', y='reviews', alpha=0.3)
plt.ylim(0, 5000)
plt.title('Price vs Number of Reviews')
plt.xlabel('Price')
plt.ylabel('Reviews')
plt.show()
```

```python
plt.figure(figsize=(8,5))
df6['isBestSeller'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['lightgray','gold'])
plt.title('Best Seller Proportion')
plt.ylabel('')
plt.show()
```

```python
display(Markdown("### 🎨 Step 6: Visual EDA"))

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

df6['categoryName'].value_counts().head(10).plot(kind='barh', ax=axes[0,0], color='steelblue')
axes[0,0].set_title('Top 10 Categories by Product Count')

axes[0,1].hist(df6['price'], bins=50, range=(0,200), color='coral')
axes[0,1].set_title('Price Distribution (under £200)')

sns.scatterplot(data=df6.sample(3000), x='price', y='reviews', alpha=0.3, ax=axes[1,0])
axes[1,0].set_ylim(0, 5000)
axes[1,0].set_title('Price vs Number of Reviews')

df6['isBestSeller'].value_counts().plot(kind='pie', ax=axes[1,1], autopct='%1.1f%%', colors=['lightgray','gold'])
axes[1,1].set_title('Best Seller Proportion')

plt.tight_layout()
plt.show()
```

```python
display(Markdown("### 🧹 Step 7: Data Cleaning"))

df6 = df6[df6['price'] > 0]

print("Shape after cleaning:", df6.shape)
print("\nMissing values now:")
print(df6.isnull().sum())
print("\nDuplicate check:", df6.duplicated().sum())
```

```python
display(Markdown("### 💾 Step 8: Save Cleaned Dataset"))

df6.to_csv(r"C:\Users\sobhi\Downloads\PRICEPIOLET AI\data\processed_data\amazon_uk_products_clean.csv", index=False)

print("Saved successfully!")
print("Final shape:", df6.shape)
```
