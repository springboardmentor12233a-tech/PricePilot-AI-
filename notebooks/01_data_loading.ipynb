```python
from pathlib import Path

import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
PROJECT_ROOT = Path("..")

RAW_DATA = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA = PROJECT_ROOT / "data" / "processed"

PROCESSED_DATA.mkdir(parents=True, exist_ok=True)

print("Raw data:", RAW_DATA)
print("Processed data:", PROCESSED_DATA)
```

```python
retail_price = pd.read_csv(RAW_DATA / "retail_price.csv")

print("Shape:", retail_price.shape)
retail_price.head()
```

```python
inventory = pd.read_csv(RAW_DATA / "retail_store_inventory.csv")

print("Shape:", inventory.shape)
inventory.head()
```

```python
# Online Retail II workbook contains separate yearly sheets, so we load both and concatenate them.
# The dataset covers UK online retail transactions from 2009–2011.

excel_file = RAW_DATA / "online_retail_II.xlsx"

excel = pd.ExcelFile(excel_file)

print("Sheets:")
print(excel.sheet_names)

online_2009_2010 = pd.read_excel(excel_file, sheet_name=excel.sheet_names[0])

online_2010_2011 = pd.read_excel(excel_file, sheet_name=excel.sheet_names[1])

online_retail = pd.concat([online_2009_2010, online_2010_2011], ignore_index=True)

print("Shape:", online_retail.shape)
online_retail.head()
```

```python
datasets = {
    "Retail Price": retail_price,
    "Inventory": inventory,
    "Online Retail": online_retail,
}

for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("Shape:", df.shape)
    print("\nColumns:")
    print(df.columns.tolist())
    print("\nData types:")
    print(df.dtypes)
```

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print(df.isnull().sum().sort_values(ascending=False).head(15))
```

```python
for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum():,} duplicate rows")
```

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name)
    display(df.describe(include="all").T)
```
