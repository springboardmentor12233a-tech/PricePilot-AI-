## 1. Dataset Loading

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("/content/sales_data.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

##2. Display the dataset

```python
df.head(10)
df.tail()
```

## 3. Dataset information

```python
df.info()
```

##4. Check missing values

```python
df.isnull().sum()
```

##5. Check duplicate records

```python
df.duplicated().sum()
```

## 6. Convert Date

```python
df["Date"] = pd.to_datetime(df["Date"])

print("Date converted successfully!")
print(df["Date"].dtype)
```

##7. Check date range

```python
print("Start date:", df["Date"].min())
print("End date:", df["Date"].max())
print("Unique dates:", df["Date"].nunique())
```

##8. Check categorical columns

```python
print("Categories:")
print(df["Category"].unique())

print("\nRegions:")
print(df["Region"].unique())

print("\nWeather Conditions:")
print(df["Weather Condition"].unique())

print("\nSeasonality:")
print(df["Seasonality"].unique())

print("\nStores:")
print(df["Store ID"].unique())

print("\nProducts:")
print(df["Product ID"].unique())
```

##9. Numerical analysis

```python
df.describe()
```

##10. Check important pricing values

```python
print("Discount values:", sorted(df["Discount"].unique()))
print("Promotion values:", sorted(df["Promotion"].unique()))
print("Epidemic values:", sorted(df["Epidemic"].unique()))
```

##11. Check negative values

```python
print("Negative Price:", (df["Price"] < 0).sum())
print("Negative Competitor Pricing:",
      (df["Competitor Pricing"] < 0).sum())
print("Negative Inventory:",
      (df["Inventory Level"] < 0).sum())
print("Negative Units Sold:",
      (df["Units Sold"] < 0).sum())
print("Negative Demand:",
      (df["Demand"] < 0).sum())
```

##12. EDA Visualization — Demand Distribution

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Demand"], bins=30)

plt.title("Demand Distribution")
plt.xlabel("Demand")
plt.ylabel("Frequency")

plt.show()
```

##13. EDA — Price Distribution

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Price"], bins=30)

plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

##14. EDA — Sales by Category

```python
category_sales = df.groupby("Category")["Units Sold"].sum()

plt.figure(figsize=(8, 5))

category_sales.plot(kind="bar")

plt.title("Units Sold by Category")
plt.xlabel("Category")
plt.ylabel("Total Units Sold")

plt.show()
```

##15. EDA — Average Demand by Category

```python
category_demand = df.groupby("Category")["Demand"].mean()

plt.figure(figsize=(8, 5))

category_demand.plot(kind="bar")

plt.title("Average Demand by Category")
plt.xlabel("Category")
plt.ylabel("Average Demand")

plt.show()
```

##16. EDA — Price vs Competitor Pricing

```python
price_comparison = df[["Price", "Competitor Pricing"]].mean()

plt.figure(figsize=(7, 5))

price_comparison.plot(kind="bar")

plt.title("Average Price vs Competitor Pricing")
plt.ylabel("Average Price")

plt.show()
```

##17. EDA — Demand over time

```python
daily_demand = df.groupby("Date")["Demand"].mean()

plt.figure(figsize=(12, 5))

plt.plot(daily_demand.index, daily_demand.values)

plt.title("Average Demand Over Time")
plt.xlabel("Date")
plt.ylabel("Average Demand")

plt.show()
```
