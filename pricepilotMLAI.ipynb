## ML Data Preparation

 In this section, we prepare the dataset for machine learning by defining the target variable, checking for data leakage, selecting relevant features, and preparing categorical and numerical variables.

```python
# Create a separate copy for machine learning
ml_df = df.copy()

# Define the target variable
target = 'Demand_Units'

print("ML dataset shape:", ml_df.shape)
print("Target variable:", target)
```

### Checking for Data Leakage

Before training the model, we identify variables that may contain information derived from the target variable. Using such variables could artificially increase model performance and result in an unrealistic model.


```python
# Check correlation of all numerical features with the target
numeric_corr = (
    ml_df.select_dtypes(include=['int64', 'float64'])
    .corr()['Demand_Units']
    .sort_values(ascending=False)
)

print(numeric_corr)
```

```python
# Inspect potentially leakage-prone variables
ml_df[
    ['Demand_Units',
     'Inventory_Turnover_Index',
     'Total_Revenue',
     'Estimated_Revenue']
].head(10)
```

```python
# Check how strongly Inventory Turnover is related to Demand
print(
    ml_df[['Demand_Units', 'Inventory_Turnover_Index']]
    .corr()
)
```

```python
# Remove leakage-prone features
leakage_features = [
    'Inventory_Turnover_Index',
    'Total_Revenue',
    'Estimated_Revenue'
]

ml_df = ml_df.drop(columns=leakage_features)

print("Removed:", leakage_features)
print("New dataset shape:", ml_df.shape)
```

```python
remaining_corr = (
    ml_df.select_dtypes(include=['int64', 'float64'])
    .corr()['Demand_Units']
    .sort_values(ascending=False)
)

print("Correlation with Demand_Units after leakage removal:")
print(remaining_corr)
```

```python
# Create useful time-based features
ml_df['Year'] = ml_df['Date'].dt.year
ml_df['Quarter'] = ml_df['Date'].dt.quarter
ml_df['DayOfWeek'] = ml_df['Date'].dt.dayofweek

print("Time features created:")
print(['Year', 'Month', 'Quarter', 'DayOfWeek'])

print("\nSample:")
print(ml_df[['Date', 'Year', 'Month', 'Quarter', 'DayOfWeek', 'Demand_Units']].head())
```

```python
# Identify categorical columns
categorical_cols = ml_df.select_dtypes(include=['object']).columns.tolist()

print("Categorical columns:")
print(categorical_cols)

print("\nUnique values in each categorical column:")
for col in categorical_cols:
    print(f"{col}: {ml_df[col].nunique()} unique values")
```

```python
product_counts = ml_df['Product_ID'].value_counts()

print("Number of unique products:", product_counts.nunique())
print("\nProduct frequency summary:")
print(product_counts.describe())

print("\nFirst 10 products and their record counts:")
print(product_counts.head(10))
```

```python
print("Current shape:", ml_df.shape)

print("\nCurrent columns:")
for i, col in enumerate(ml_df.columns, start=1):
    print(f"{i}. {col}")
```

```python
# Remove raw Date because we already extracted useful time features
ml_df = ml_df.drop(columns=['Date'])

print("Raw Date column removed.")
print("Current shape:", ml_df.shape)
```

```python
# Define features and target
X = ml_df.drop(columns=['Demand_Units'])
y = ml_df['Demand_Units']

print("Features (X) shape:", X.shape)
print("Target (y) shape:", y.shape)

print("\nTarget variable:")
print(y.name)
```

```python
# Identify categorical and numerical features
categorical_features = X.select_dtypes(include=['object']).columns.tolist()
numerical_features = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)

print("\nNumber of categorical features:", len(categorical_features))
print("Number of numerical features:", len(numerical_features))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_features
        ),
        (
            'numerical',
            'passthrough',
            numerical_features
        )
    ]
)

print("Preprocessing pipeline created successfully.")
```

```python
print(ml_df['Year'].value_counts().sort_index())
```

```python
# Time-based train-test split
train_mask = ml_df['Year'] < 2025
test_mask = ml_df['Year'] == 2025

X_train = X[train_mask]
X_test = X[test_mask]

y_train = y[train_mask]
y_test = y[test_mask]

print("Training set:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting set:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
# Fit preprocessing only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Apply the same transformation to test data
X_test_processed = preprocessor.transform(X_test)

print("Training data after preprocessing:", X_train_processed.shape)
print("Testing data after preprocessing:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LinearRegression

# Create the baseline model
linear_model = LinearRegression()

# Train the model
linear_model.fit(X_train_processed, y_train)

print("Linear Regression model trained successfully.")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Make predictions on the test set
y_pred_linear = linear_model.predict(X_test_processed)

# Calculate evaluation metrics
mae_linear = mean_absolute_error(y_test, y_pred_linear)
rmse_linear = np.sqrt(mean_squared_error(y_test, y_pred_linear))
r2_linear = r2_score(y_test, y_pred_linear)

print("Linear Regression Performance")
print("-" * 35)
print(f"MAE  : {mae_linear:.4f}")
print(f"RMSE : {rmse_linear:.4f}")
print(f"R²   : {r2_linear:.4f}")
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_processed, y_train)

print("Random Forest training completed.")
```

```python
def recommend_price(sample_row):
    sample = sample_row.copy()

    competitor_price = sample["Competitor_Price"]

    min_price = competitor_price * 0.80
    max_price = competitor_price * 1.20

    candidate_prices = np.linspace(
        min_price,
        max_price,
        9
    )

    results = []

    for price in candidate_prices:

        scenario = sample.copy()

        scenario["Price"] = price
        scenario["Price_Difference"] = price - competitor_price
        scenario["Price_Ratio"] = price / competitor_price

        if price < competitor_price:
            scenario["Price_Position"] = "Cheaper"
        else:
            scenario["Price_Position"] = "More Expensive"

        scenario_df = pd.DataFrame([scenario])

        scenario_processed = preprocessor.transform(scenario_df)

        predicted_demand = xgb_model.predict(
            scenario_processed
        )[0]

        expected_revenue = price * predicted_demand

        results.append({
            "Candidate_Price": round(price, 2),
            "Predicted_Demand": round(predicted_demand, 2),
            "Expected_Revenue": round(expected_revenue, 2)
        })

    results_df = pd.DataFrame(results)

    best = results_df.loc[
        results_df["Expected_Revenue"].idxmax()
    ]

    return best, results_df
```

```python
best_price, pricing_results = recommend_price(X_test.iloc[0])

print("PricePilot Recommendation")
print("-" * 40)
print(f"Recommended Price : ₹{best_price['Candidate_Price']:.2f}")
print(f"Predicted Demand  : {best_price['Predicted_Demand']:.2f} units")
print(f"Expected Revenue  : ₹{best_price['Expected_Revenue']:.2f}")
```

```python
# Step 1: Prepare daily demand data for forecasting

forecast_df = df.copy()

# Convert Date to datetime
forecast_df['Date'] = pd.to_datetime(forecast_df['Date'])

# Aggregate total demand by date
daily_demand = (
    forecast_df
    .groupby('Date')['Demand_Units']
    .sum()
    .reset_index()
    .sort_values('Date')
)

print("Daily demand data prepared successfully.")
print("Number of days:", len(daily_demand))

print("\nFirst 5 rows:")
print(daily_demand.head())

print("\nLast 5 rows:")
print(daily_demand.tail())
```

```python
# Step 3: Create monthly demand data

monthly_demand = (
    daily_demand
    .set_index('Date')
    .resample('ME')['Demand_Units']
    .sum()
    .reset_index()
)

print("Monthly demand data prepared successfully.")
print("Number of months:", len(monthly_demand))

print("\nFirst 5 months:")
print(monthly_demand.head())

print("\nLast 5 months:")
print(monthly_demand.tail())
```

```python
# Step 5: Time-based train-test split for demand forecasting

train_forecast = monthly_demand[
    monthly_demand['Date'].dt.year < 2025
].copy()

test_forecast = monthly_demand[
    monthly_demand['Date'].dt.year == 2025
].copy()

print("Forecasting train data:", train_forecast.shape)
print("Forecasting test data:", test_forecast.shape)

print("\nTraining period:")
print(train_forecast['Date'].min(), "to", train_forecast['Date'].max())

print("\nTesting period:")
print(test_forecast['Date'].min(), "to", test_forecast['Date'].max())
```

```python
# Step 6: Create lag features for demand forecasting

forecast_features = monthly_demand.copy()

forecast_features['Lag_1'] = forecast_features['Demand_Units'].shift(1)
forecast_features['Lag_2'] = forecast_features['Demand_Units'].shift(2)
forecast_features['Lag_3'] = forecast_features['Demand_Units'].shift(3)
forecast_features['Lag_6'] = forecast_features['Demand_Units'].shift(6)
forecast_features['Lag_12'] = forecast_features['Demand_Units'].shift(12)

# Add time-based features
forecast_features['Month'] = forecast_features['Date'].dt.month
forecast_features['Year'] = forecast_features['Date'].dt.year

print("Lag features created successfully.")
print("\nFirst 15 rows:")
print(forecast_features.head(15))
```

```python
# Step 7: Remove rows with missing lag values

forecast_features = forecast_features.dropna().reset_index(drop=True)

print("Forecasting dataset prepared.")
print("Shape:", forecast_features.shape)

print("\nFirst 5 rows:")
print(forecast_features.head())

print("\nMissing values:")
print(forecast_features.isnull().sum())
```

```python
# Step 8: Create train-test split for forecasting

forecast_train = forecast_features[
    forecast_features['Year'] < 2025
].copy()

forecast_test = forecast_features[
    forecast_features['Year'] == 2025
].copy()

# Features used by the forecasting model
forecast_X_train = forecast_train[
    ['Lag_1', 'Lag_2', 'Lag_3', 'Lag_6', 'Lag_12', 'Month']
]

forecast_y_train = forecast_train['Demand_Units']

forecast_X_test = forecast_test[
    ['Lag_1', 'Lag_2', 'Lag_3', 'Lag_6', 'Lag_12', 'Month']
]

forecast_y_test = forecast_test['Demand_Units']

print("Forecast training data:", forecast_X_train.shape)
print("Forecast testing data:", forecast_X_test.shape)

print("\nTraining period:")
print(forecast_train['Date'].min(), "to", forecast_train['Date'].max())

print("\nTesting period:")
print(forecast_test['Date'].min(), "to", forecast_test['Date'].max())
```

```python
# Step 9: Train XGBoost demand forecasting model

from xgboost import XGBRegressor

forecast_model = XGBRegressor(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=3,
    random_state=42,
    n_jobs=-1,
    objective='reg:squarederror'
)

forecast_model.fit(
    forecast_X_train,
    forecast_y_train
)

print("Demand forecasting model trained successfully.")
```

```python
# Step 10: Generate 2025 demand forecasts

forecast_test['Predicted_Demand'] = forecast_model.predict(
    forecast_X_test
)

print("2025 demand forecasts generated successfully.")

print("\nActual vs Predicted Demand:")
print(
    forecast_test[
        ['Date', 'Demand_Units', 'Predicted_Demand']
    ].to_string(index=False)
)
```

```python
# Step 11: Evaluate demand forecasting model

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

forecast_actual = forecast_test['Demand_Units']
forecast_predicted = forecast_test['Predicted_Demand']

forecast_mae = mean_absolute_error(
    forecast_actual,
    forecast_predicted
)

forecast_rmse = np.sqrt(
    mean_squared_error(
        forecast_actual,
        forecast_predicted
    )
)

forecast_r2 = r2_score(
    forecast_actual,
    forecast_predicted
)

print("Demand Forecasting Model Performance on 2025 Data")
print("=" * 55)

print(f"MAE  : {forecast_mae:.2f}")
print(f"RMSE : {forecast_rmse:.2f}")
print(f"R²   : {forecast_r2:.4f}")
```

```python
# Step 13: Add rolling and seasonal features for forecasting

forecast_enhanced = monthly_demand.copy()

# Previous-month demand
forecast_enhanced['Lag_1'] = forecast_enhanced['Demand_Units'].shift(1)

# Previous 3-month demand
forecast_enhanced['Lag_3'] = forecast_enhanced['Demand_Units'].shift(3)

# Same month from previous year
forecast_enhanced['Lag_12'] = forecast_enhanced['Demand_Units'].shift(12)

# Rolling average of previous 3 months
forecast_enhanced['Rolling_Mean_3'] = (
    forecast_enhanced['Demand_Units']
    .shift(1)
    .rolling(3)
    .mean()
)

# Rolling average of previous 6 months
forecast_enhanced['Rolling_Mean_6'] = (
    forecast_enhanced['Demand_Units']
    .shift(1)
    .rolling(6)
    .mean()
)

# Time features
forecast_enhanced['Month'] = forecast_enhanced['Date'].dt.month
forecast_enhanced['Year'] = forecast_enhanced['Date'].dt.year

# Remove rows with missing lag/rolling values
forecast_enhanced = forecast_enhanced.dropna().reset_index(drop=True)

print("Enhanced forecasting dataset created.")
print("Shape:", forecast_enhanced.shape)
print("\nColumns:")
print(forecast_enhanced.columns.tolist())

print("\nFirst 5 rows:")
print(forecast_enhanced.head())
```
